```python
import os
import pandas as pd
import _common, _map, _plot, _dist
device = 'SM-S938N'
out_path = 'results'
grid_size = 30
sample_min = 30
rb_min = 0
rsrp_bin = 1

df = _common.analyze_kpi(
    fname = os.path.join("logs",f"{device}.xlsx"),
    date_list = [
        # "251016",
        # "251017",
        # "251020",
        # "251021",
        "251022", # pdcch dynamic 적용
        "251023",
        "251024",
        "251027",
        "251103",
        "251105",
    ],
)
df_map = df[df["route"].isin(['Namsan','Huam415-1','Huam345-5'])].copy()
print()

_common.grid_kpi(device=device, df=df_map, rb_min=rb_min, grid_size=grid_size, sample_min=sample_min)
print()

df_grid = pd.read_excel(os.path.join("logs", f"{device}_grid_{grid_size}m.xlsx"))

out_dir = os.path.join(out_path, "map")

_map.map_coverage(df=df_grid, out_dir=out_dir, grid_size=grid_size, band="n28")
_map.map_uhd_pwr(df=df_grid, out_dir=out_dir, grid_size=grid_size)
_map.map_pct(df=df_grid, out_dir=out_dir, grid_size=grid_size)
_map.map_db(df=df_grid, out_dir=out_dir, grid_size=grid_size)


out_dir = os.path.join(out_path, "dist", "uhd_pwr")

_dist.dist_uhd_by_minmax(df=df_grid, out_dir=out_dir)
_dist.dist_uhd_by_site(df=df_grid, out_dir=out_dir, grid_size=grid_size)
```

```python
import os
import pandas as pd
import _common, _map, _plot, _dist
device = 'SM-S938N'
out_path = 'results'
grid_size = 30
sample_min = 30
rb_min = 0
rsrp_bin = 1

df = _common.analyze_kpi(
    fname = os.path.join("logs",f"{device}.xlsx"),
    date_list = [
        # "251016",
        # "251017",
        # "251020",
        # "251021",
        "251022", # pdcch dynamic 적용
        "251023",
        "251024",
        "251027",
        "251103",
        "251105",
    ],
)

out_dir = os.path.join(out_path, "plot")

_plot.plot_kpis_by_site(df=df, out_dir=out_dir, rb_min=rb_min, sample_min=sample_min, rsrp_bin=rsrp_bin)
_plot.plot_kpis_by_band(df=df, out_dir=out_dir, rb_min=rb_min, sample_min=sample_min, rsrp_bin=rsrp_bin)
_plot.plot_kpis_by_uhd(df=df, out_dir=out_dir, rb_min=rb_min, sample_min=sample_min, rsrp_bin=rsrp_bin, grid_size=grid_size)


out_dir = os.path.join(out_path, "plot_fixed_point")

_plot.plot_fixed_point(df=df, out_dir=out_dir, rb_min=rb_min, sample_min=sample_min, rsrp_bin=rsrp_bin, grid_size=grid_size)
_plot.plot_fixed_point_apple(df=df, out_dir=out_dir, rb_min=rb_min, sample_min=sample_min, rsrp_bin=rsrp_bin, grid_size=grid_size)


out_dir = os.path.join(out_path, "dist")

_dist.dist_kpis_pdf_rsrp_bins(df=df, out_dir=out_dir, rb_min=rb_min, rsrp_bin=rsrp_bin)

_dist.dist_kpis_pdf_by_band(df=df, out_dir=out_dir, rb_min=rb_min)
_dist.dist_kpis_pdf_by_site(df=df, out_dir=out_dir, rb_min=rb_min)
_dist.dist_kpis_pdf_by_uhd(df=df, out_dir=out_dir, rb_min=rb_min, grid_size=grid_size, sample_min=sample_min)

# _dist.dist_kpis_cdf_by_band(df=df, out_dir=out_dir, rb_min=rb_min)
# _dist.dist_kpis_cdf_by_site(df=df, out_dir=out_dir, rb_min=rb_min)
# _dist.dist_kpis_cdf_by_uhd(df=df, out_dir=out_dir, rb_min=rb_min, grid_size=grid_size)
```
